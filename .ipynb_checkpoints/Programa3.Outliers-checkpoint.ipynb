# 0.1 Detección de Outliers

```python
# ==========================================================
# Importación de librerías
# ==========================================================

# Importamos pandas para trabajar con el conjunto de datos
# en forma de DataFrame.
import pandas as pd

# Importamos NumPy para realizar operaciones numéricas
# y trabajar con posiciones dentro de los datos.
import numpy as np

# Importamos las herramientas estadísticas de SciPy,
# que utilizaremos posteriormente para el método Box Plot.
from scipy import stats

# Importamos EllipticEnvelope, algoritmo utilizado
# para detectar valores atípicos o outliers.
from sklearn.covariance import EllipticEnvelope

# Importamos Matplotlib para generar las visualizaciones.
import matplotlib.pyplot as plt
```

```python
# ==========================================================
# Carga de datos
# ==========================================================

# Leemos el archivo CSV que se encuentra dentro de la
# carpeta "dataset".
df = pd.read_csv("dataset/outliers.csv")

# Mostramos el contenido del DataFrame para observar
# las variables y los registros disponibles.
print(df)
```

0.1.1 EllipticEnvelope()

```python
# ==========================================================
# 1) Detección de outliers mediante EllipticEnvelope
# ==========================================================

# Creamos el algoritmo EllipticEnvelope.
# contamination=0.25 indica que se considera aproximadamente
# un 25% de los datos como posibles outliers.
# random_state=42 permite obtener resultados reproducibles.
algorithm = EllipticEnvelope(
    support_fraction=None,
    contamination=0.25,
    random_state=42
)

# Ajustamos el algoritmo utilizando nuestro DataFrame.
outlier_method = algorithm.fit(df)

# Aplicamos el método entrenado sobre el conjunto de datos.
# El resultado identifica cada registro como:
#  1  -> dato considerado normal
# -1  -> dato considerado outlier
df_outliers = outlier_method.predict(df)

# Mostramos la clasificación obtenida para cada registro.
print(df_outliers)
```

```python
# ==========================================================
# Determinar la posición de los outliers
# ==========================================================

# np.where() permite encontrar las posiciones donde
# el algoritmo ha asignado el valor -1.
pos_outliers = np.where(df_outliers == -1)[0]

# Mostramos las posiciones de los registros considerados
# como outliers.
print('\nOutliers en la posición: \n', pos_outliers)

# Contamos cuántos registros fueron detectados como outliers.
print('\nNúmero de outliers: \n', len(pos_outliers))
```

```python
# ==========================================================
# Visualizamos las primeras 10 filas del dataset
# ==========================================================

# Mostramos únicamente las primeras 10 filas
# para observar los datos con los que estamos trabajando.
df.head(10)
```

```python
# ==========================================================
# Otros métodos de detección de outliers
# ==========================================================

# Importamos diferentes algoritmos para detectar outliers.
from sklearn.ensemble import IsolationForest
from sklearn.svm import OneClassSVM
from sklearn.neighbors import LocalOutlierFactor

# Creamos el modelo Isolation Forest.
# max_samples='auto' permite que el algoritmo determine
# automáticamente el número de muestras utilizadas.
# random_state=42 permite reproducir los resultados.
IF = IsolationForest(
    max_samples='auto',
    random_state=42
)

# Estas dos líneas aparecen como ejercicios pendientes
# en el documento original.
# OC_SVM = ??
# LOF = ??

# Aplicamos Isolation Forest utilizando la función
# find_outliers() creada anteriormente.
df_outliers, pos_outliers = find_outliers(df, IF)

# Mostramos la cantidad de outliers encontrados.
print(len(pos_outliers))
```

```python
# ==========================================================
# Visualización de la frontera de decisión
# ==========================================================

# Importamos la herramienta para visualizar
# la frontera de decisión del modelo.
from sklearn.inspection import DecisionBoundaryDisplay

# Generamos la frontera de decisión de Isolation Forest.
disp = DecisionBoundaryDisplay.from_estimator(
    IF,
    df,
    response_method="decision_function",
    alpha=0.5,
)

# Dibujamos los puntos del dataset.
#  1 representa datos normales.
# -1 representa outliers.
disp.ax_.scatter(
    df.a,
    df.b,
    c=df_outliers,
    s=20,
    edgecolor="k"
)

# Colocamos el título del gráfico.
disp.ax_.set_title(
    "Binary Decision boundary \nof IsolationForest"
)

# Igualamos la escala de los ejes.
plt.axis("square")

# Agregamos la leyenda.
plt.legend(
    labels=["outliers", "inliers"],
    title="true class"
)

# Mostramos el gráfico.
plt.show()
```

```python
# ==========================================================
# Función para detectar outliers
# ==========================================================

# Creamos una función que recibe:
# - df: conjunto de datos
# - algorithm: algoritmo utilizado para detectar outliers
def find_outliers(df, algorithm):

    # Ajustamos el algoritmo utilizando los datos recibidos.
    outlier_method = algorithm.fit(df)

    # Clasificamos cada registro.
    #  1  representa un dato normal.
    # -1 representa un outlier.
    df_outliers = outlier_method.predict(df)

    # Buscamos las posiciones donde se encontraron outliers.
    pos_outliers = np.where(df_outliers == -1)[0]

    # Mostramos las posiciones de los outliers.
    print('\nOutliers en la posición: \n', pos_outliers)

    # Mostramos el número total de outliers encontrados.
    print('\nNúmero de outliers: \n', len(pos_outliers))

    # Devolvemos las posiciones para poder utilizarlas
    # posteriormente si fuera necesario.
    return pos_outliers
```

```python
# ==========================================================
# Configuración de Isolation Forest
# ==========================================================

# Creamos el algoritmo IsolationForest.
# contamination=0.25 indica que aproximadamente el 25%
# de los datos pueden ser considerados outliers.
# random_state=42 permite reproducir los resultados.
algorithm = IsolationForest(
    contamination=0.25,
    random_state=42
)
```

```python
# ==========================================================
# Detección de outliers con Isolation Forest
# ==========================================================

# Utilizamos la función creada anteriormente para ajustar
# el algoritmo y encontrar las posiciones de los outliers.
pos_outliers = find_outliers(df, algorithm)
```

0.1.3 — 3) Box plot

```python
# ==========================================================
# 3) Box Plot
# ==========================================================

# Seleccionamos el atributo que vamos a analizar.
a = df['a']

# Calculamos el primer cuartil (Q1), que corresponde
# al percentil 25 de los datos.
Q1 = stats.scoreatpercentile(a, 25)

# Calculamos el tercer cuartil (Q3), que corresponde
# al percentil 75 de los datos.
Q3 = stats.scoreatpercentile(a, 75)

# Calculamos el rango intercuartílico (RIC).
# El RIC representa la diferencia entre Q3 y Q1.
RIC = Q3 - Q1

# Calculamos el límite inferior para identificar outliers.
li = Q1 - 1.5 * RIC

# Calculamos el límite superior para identificar outliers.
ls = Q3 + 1.5 * RIC

# Mostramos los límites calculados.
print('limite inferior: ', li)
print('limite superior: ', ls)

# Buscamos las posiciones de los valores inferiores
# al límite inferior.
pos_i = np.where(a < li)[0]

# Buscamos las posiciones de los valores superiores
# al límite superior.
pos_s = np.where(a > ls)[0]

# Unimos las posiciones de ambos tipos de outliers.
pos_outliers = np.concatenate((pos_i, pos_s))

# Mostramos las posiciones encontradas.
print('Posición de outliers: ', pos_outliers)

# Mostramos cuántos outliers fueron detectados.
print('Número de outliers: ', len(pos_outliers))

# Dibujamos el diagrama de caja y bigotes.
prop = plt.boxplot(a)

# Mostramos nuevamente el Box Plot.
plt.boxplot(a)

# Mostramos el gráfico.
plt.show()
```

```python
# ==========================================================
# Función para detectar outliers mediante Box Plot
# ==========================================================

# Definimos una función que recibe una columna
# del DataFrame y devuelve las posiciones de sus outliers.
def find_limits_BP(variable):

    # Calculamos el primer cuartil de la variable.
    Q1 = stats.scoreatpercentile(variable, 25)

    # Calculamos el tercer cuartil de la variable.
    Q3 = stats.scoreatpercentile(variable, 75)

    # Calculamos el rango intercuartílico.
    RIC = Q3 - Q1

    # Calculamos el límite inferior.
    li = Q1 - 1.5 * RIC

    # Calculamos el límite superior.
    ls = Q3 + 1.5 * RIC

    # Obtenemos las posiciones de los valores
    # que están por debajo del límite inferior.
    pos_i = np.where(variable < li)[0]

    # Obtenemos las posiciones de los valores
    # que están por encima del límite superior.
    pos_s = np.where(variable > ls)[0]

    # Unimos las posiciones de ambos grupos.
    pos_outliers = np.concatenate((pos_i, pos_s))

    # Devolvemos las posiciones encontradas.
    return pos_outliers
```

```python
# ==========================================================
# Detección de outliers en todos los atributos
# ==========================================================

# Obtenemos los nombres de todas las columnas del DataFrame.
headers = df.columns

# Creamos una lista donde almacenaremos las posiciones
# de los outliers encontrados en cada atributo.
pos_outliers = []

# Recorremos todas las columnas del DataFrame.
for i in range(len(headers)):

    # Seleccionamos el atributo actual.
    variable = df[headers[i]]

    # Aplicamos la función de Box Plot que creamos
    # anteriormente para encontrar sus outliers.
    pos_out = np.expand_dims(
        find_limits_BP(variable),
        axis=1
    )

    # Guardamos las posiciones encontradas.
    pos_outliers.append(pos_out)

# Concatenamos las posiciones de los outliers
# encontrados en todas las variables.
po = np.vstack(pos_outliers)

# Eliminamos las posiciones repetidas.
pos_out = np.unique(po)

# Mostramos las posiciones de todos los outliers.
print('Posiciones de outliers: ', pos_out)

# Mostramos el número total de posiciones de outliers.
print('Numero de outliers: ', len(pos_out))
```
