# Caso Práctico: Neurona de McCulloch y Pitts

## 2. Aplicando la MP Neurona a un caso práctico real

### 2.1. Conjunto de datos

En esta práctica se utiliza el conjunto de datos Breast Cancer Wisconsin, disponible en la biblioteca `scikit-learn`.

Las características de entrada se calculan a partir de imágenes digitalizadas de muestras obtenidas mediante aspiración con aguja fina (FNA). Estas características describen propiedades de los núcleos celulares presentes en las imágenes.

El conjunto de datos permite trabajar con dos clases: tumores benignos y malignos. Primero se revisará la estructura de los datos y después se prepararán las características para utilizarlas con una neurona MP.

```python

# Importamos el conjunto de datos de cáncer de mama de sklearn.
from sklearn.datasets import load_breast_cancer

# Cargamos la información en una variable.
cancer_mama = load_breast_cancer()

# Separamos las características de entrada y las etiquetas de salida.
X = cancer_mama.data
Y = cancer_mama.target
```

```python

# Mostramos los atributos que contiene el conjunto de datos.
dir(cancer_mama)
```

```python

# Visualizamos las características numéricas de las muestras.
print(X)
```


## Visualización del conjunto de datos

Para facilitar la interpretación de la información, convertimos las características en un DataFrame de pandas. Cada columna representa una característica de las muestras y cada fila corresponde a un registro.

Después observamos las etiquetas de salida, que indican la clase asignada a cada muestra.

```python

# Importamos pandas para organizar los datos en una tabla.
import pandas as pd

# Creamos el DataFrame utilizando los nombres originales
# de las características del conjunto de datos.
df = pd.DataFrame(X, columns=cancer_mama.feature_names)

# Mostramos la tabla para revisar su contenido.
df
```

```python

# Mostramos las etiquetas que indican la clase de cada muestra.
print(Y)
```


## División del conjunto de datos

Dividimos las características y sus etiquetas en conjuntos de entrenamiento y prueba. El primero se utiliza para ajustar el modelo y el segundo para evaluar sus predicciones.

El parámetro `stratify=Y` ayuda a conservar la proporción de las clases en ambos conjuntos. De esta manera, la evaluación representa mejor la distribución original de las etiquetas.

```python

# Importamos la función para dividir los datos.
from sklearn.model_selection import train_test_split

# Separamos los datos en entrenamiento y prueba,
# conservando la proporción de las clases.
X_train, X_test, y_train, y_test = train_test_split(
    df, Y, stratify=Y
)

# Mostramos el número de registros de cada conjunto.
print("Tamaño de entrenamiento:", len(X_train))
print("Tamaño de pruebas:", len(X_test))
```


## Implementación de una MP Neurona más avanzada

En esta sección creamos la clase `MpNeuron`, que contiene los métodos necesarios para realizar predicciones y ajustar el umbral de la neurona.

- `__init__`: inicializa el umbral.
- `model`: suma las entradas y compara el resultado con el umbral.
- `predict`: realiza predicciones para varias muestras.
- `fit`: prueba diferentes umbrales y selecciona el que obtiene la mayor exactitud sobre los datos de entrenamiento.

La neurona MP trabaja con entradas binarias, por lo que será necesario transformar las características numéricas antes de entrenarla.

```python

# Importamos NumPy para trabajar con arreglos.
import numpy as np

# Importamos la función que calcula la exactitud del modelo.
from sklearn.metrics import accuracy_score


class MpNeuron:
    # Inicializamos la neurona sin un umbral definido.
    def __init__(self):
        self.threshold = None

    # Sumamos las entradas y comparamos el resultado con el umbral.
    def model(self, x):
        # Las entradas son valores binarios: 0 o 1.
        z = sum(x)

        # Devolvemos True si la suma alcanza el umbral.
        return (z >= self.threshold)

    # Realizamos predicciones para un conjunto de muestras.
    def predict(self, X):
        Y = []

        # Recorremos cada muestra de entrada.
        for x in X:
            result = self.model(x)
            Y.append(result)

        # Convertimos las predicciones en un arreglo de NumPy.
        return np.array(Y)

    # Buscamos el umbral que consigue la mejor exactitud.
    def fit(self, X, Y):
        accuracy = {}

        # Probamos los posibles umbrales según el número
        # de características de entrada.
        for th in range(X.shape[1] + 1):
            self.threshold = th

            # Calculamos las predicciones con el umbral actual.
            Y_pred = self.predict(X)

            # Guardamos la exactitud obtenida.
            accuracy[th] = accuracy_score(Y_pred, Y)

        # Seleccionamos el umbral con mayor exactitud.
        self.threshold = max(accuracy, key=accuracy.get)
```


### Transformación de los datos a valores binarios

Las características del conjunto de datos contienen valores numéricos continuos. Sin embargo, la neurona MP de esta práctica utiliza entradas binarias.

Para adaptar los datos, utilizamos `pd.cut` con dos intervalos y etiquetas de 1 y 0. También representamos un ejemplo mediante un histograma para observar cómo se distribuyen los valores en los intervalos.

```python

# Importamos Matplotlib para crear el histograma.
import matplotlib.pyplot as plt

# Dividimos estos valores de ejemplo en dos intervalos
# y asignamos las etiquetas binarias 0 y 1.
valores = [0.04, 0.3, 4, 5, 6, 0.02, 0.6]

print(pd.cut(valores, bins=2, labels=[0, 1]))

# Representamos la distribución de los valores.
plt.hist(valores, bins=2)
plt.show()
```

```python

# Transformamos las características de entrenamiento a valores binarios.
X_train_bin = X_train.apply(
    pd.cut, bins=2, labels=[1, 0]
)

# Aplicamos la misma transformación a los datos de prueba.
X_test_bin = X_test.apply(
    pd.cut, bins=2, labels=[1, 0]
)

# Mostramos los datos binarizados de entrenamiento.
X_train_bin
```


## Entrenamiento de la neurona MP

En esta etapa creamos el modelo y lo entrenamos con las características binarias y sus etiquetas correspondientes.

El método `fit` evalúa distintos valores de umbral y conserva el que logra la mayor exactitud en el conjunto de entrenamiento. Después mostramos el umbral seleccionado.

```python

# Creamos una instancia de la neurona MP.
np_neurona = MpNeuron()

# Entrenamos el modelo con los datos binarios.
np_neurona.fit(X_train_bin.to_numpy(), y_train)

# Mostramos el umbral óptimo seleccionado.
np_neurona.threshold
```


## Predicción y evaluación del modelo

En esta sección usamos la neurona entrenada para generar predicciones sobre el conjunto de prueba.

Después calculamos la exactitud comparando las predicciones con las etiquetas reales. Esta métrica permite observar qué proporción de las muestras fue clasificada correctamente.

```python

# Realizamos las predicciones con las características binarias de prueba.
Y_pred = np_neurona.predict(X_test_bin.to_numpy())
```

```python

# Mostramos las predicciones obtenidas por la neurona.
Y_pred
```

```python

# Calculamos la exactitud comparando las clases reales
# con las clases predichas por el modelo.
accuracy_score(y_test, Y_pred)
```
