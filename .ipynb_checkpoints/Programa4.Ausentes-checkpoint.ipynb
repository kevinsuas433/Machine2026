```python
# Importamos NumPy para trabajar con valores numéricos
# y representar los datos ausentes mediante np.nan.
import numpy as np

# Importamos SimpleImputer de Scikit-learn.
# Esta herramienta permite reemplazar valores ausentes
# utilizando diferentes estrategias, como la media.
from sklearn.impute import SimpleImputer
```

```python
# Creamos el conjunto de datos de entrenamiento.
# El valor np.nan representa un dato ausente.
X_train = [
    [1, 4],
    [np.nan, 3],
    [7, 8]
]

# Mostramos los datos originales para observar
# dónde se encuentra el valor ausente.
print(X_train)
```

```python
# Creamos el objeto de imputación.
# Utilizamos la estrategia 'mean', que reemplaza
# cada valor ausente por la media de su columna.
imputer = SimpleImputer(strategy='mean')

# Ajustamos el imputador utilizando los datos de entrenamiento
# y transformamos los valores ausentes.
X_train_clean = imputer.fit_transform(X_train)

# Mostramos el conjunto de datos después de la imputación.
print(X_train_clean)
```

```python
# Creamos el conjunto de datos de prueba.
# En este caso no tenemos valores ausentes.
X_test = [[np.nan, 10],
[6, np.nan],
[8, 2]]


# Aplicamos el mismo imputador que fue ajustado
# anteriormente con los datos de entrenamiento.
# Es importante utilizar transform() y no fit_transform(),
# porque el imputador ya aprendió las medias de X_train.
X_test_clean = imputer.transform(X_test)

# Mostramos los datos después del proceso de imputación.
print(X_test_clean)
```
