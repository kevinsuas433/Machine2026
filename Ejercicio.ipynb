```python
#Importamos libreeias
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt 
```

```python
#Creamos el conjunto de datos 
df = pd.DataFrame ({
    "horas": [1,2,3,4,5],
    "nota": [2,3,5,4,6]
})
x= df[["horas"]]
y= df[["nota"]]
print (x)
```

```python
#Creamos nuestro modelo 
modelo = LinearRegression()
modelo.fit(x, y)
```

```python
#Imprimimos el interceptor y el coef 
print("Interceptor: ", modelo.intercept_)
print("Pendiente:", modelo.coef_[0])
```

```python
#Valores predichos
y_pred = modelo.predict(x)
```

```python
#Realizamos la nueva prediccion 7
horas=float(input("Ingrese las horas de estudio"))
prediccion=modelo.predict([[horas]])
print("Prediccion:", prediccion)
```

# GRAFICO 

```python
import matplotlib.pyplot as plt
import numpy as np

# Datos del conjunto original
x_datos = np.array(x).flatten()
y_reales = np.array(y).flatten()
y_predichos = np.array(y_pred).flatten()

# Ordenamos X para dibujar la pendiente
orden = np.argsort(x_datos)
x_ordenado = x_datos[orden]

# Línea de regresión usando los valores que YA tiene el modelo
y_linea = modelo.intercept_[0] + modelo.coef_[0] * x_ordenado

# ==============================
# GRÁFICO
# ==============================

plt.figure(figsize=(10, 6))

# Datos reales del conjunto
plt.scatter(
    x_datos,
    y_reales,
    color='blue',
    label='Datos reales'
)

# Datos predichos
plt.scatter(
    x_datos,
    y_predichos,
    color='red',
    label='Datos predichos'
)

# Dato nuevo ingresado: 6 horas
plt.scatter(
    horas,
    prediccion[0],
    color='green',
    s=120,
    label='Dato ingresado'
)

# Pendiente / línea de regresión
plt.plot(
    x_ordenado,
    y_linea,
    color='black',
    linewidth=2,
    label='Línea de regresión'
)

plt.xlabel('Horas de estudio')
plt.ylabel('Calificación')
plt.title('Regresión Lineal')
plt.legend()
plt.grid(True)

plt.show()
```
