```python
# Importamos NumPy para trabajar con arreglos y operaciones numéricas.
import numpy as np

# Importamos VarianceThreshold de scikit-learn.
# Esta herramienta permite seleccionar atributos según su varianza,
# eliminando aquellos que presentan poca variación.
from sklearn.feature_selection import VarianceThreshold
```

```python
# Definimos el conjunto de datos de entrada.
# Cada lista representa una observación y cada posición
# corresponde a un atributo o característica.
X = [
    [0, 0, 1],
    [0, 1, 0],
    [1, 0, 0],
    [0, 1, 1],
    [0, 1, 0],
    [0, 1, 1]
]

# Mostramos los datos cargados.
print(X)
```

```python
# Definimos el umbral mínimo de varianza.
# Los atributos cuya variación sea menor que este valor
# serán eliminados del conjunto de datos.
th = 0.2

# Creamos el selector utilizando el umbral establecido.
sel = VarianceThreshold(th)

# Ajustamos el selector a los datos y transformamos la matriz,
# conservando únicamente los atributos que superan el umbral.
print(sel.fit_transform(X))
```
